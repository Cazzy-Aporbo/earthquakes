```python
import requests
import pandas as pd

# Define the API endpoint with parameters
api_url = 'https://earthquake.usgs.gov/fdsnws/event/1/query?format=geojson&starttime=1900-01-01&endtime=2024-03-06&minmagnitude=7'

# Make the request to the USGS API
response = requests.get(api_url)

# Check if the request was successful
if response.status_code == 200:
    # Parse the JSON response
    data = response.json()
    
    # Extract the earthquake data
    earthquakes = data['features']
    
    # Convert to DataFrame
    df = pd.json_normalize(earthquakes)
    
    # Save to CSV file
    csv_filename = 'earthquake_data.csv'
    df.to_csv(csv_filename, index=False)
    print(f'Data saved to {csv_filename}')
else:
    print('Failed to retrieve data')
```
